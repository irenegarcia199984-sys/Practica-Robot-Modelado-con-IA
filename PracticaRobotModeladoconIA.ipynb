```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

# ==========================================
# 1. BASE DE CONOCIMIENTOS DEL ROBOT
# ==========================================
# Matriz X: Representa las lecturas de los sensores en metros [Izq, Centro, Der]
X = np.array([
    [1.5, 2.0, 1.5],  # Fila 0: Camino libre -> Avanzar (0)
    [0.2, 0.5, 1.5],  # Fila 1: Obstáculo a izquierda/centro -> Girar Derecha (2)
    [1.5, 0.5, 0.2],  # Fila 2: Obstáculo a derecha/centro -> Girar Izquierda (1)
    [0.1, 0.1, 1.5],  # Fila 3: Obstáculo frontal crítico -> Girar Derecha (2)
])

# Vector y: Etiquetas con la decisión lógica correcta para cada caso
y = np.array([0, 2, 1, 2])

# Diccionario para traducir las decisiones numéricas a comandos legibles
acciones = {
    0: "Avanzar 🟢",
    1: "Girar a la izquierda ↩️",
    2: "Girar a la derecha ↪️"
}

# ==========================================
# 2. ENTRENAMIENTO DEL MODELO DE IA (KNN)
# ==========================================
# Inicializamos el clasificador K-Nearest Neighbors con k=1
# para que el robot imite el patrón conocido más similar.
knn = KNeighborsClassifier(n_neighbors=1)

# Entrenamos el modelo asociando las mediciones de los sensores con sus acciones
knn.fit(X, y)

# ==========================================
# 3. SIMULACIÓN DEL ROBOT EN TIEMPO REAL
# ==========================================
print("=== SIMULACIÓN DEL ROBOT AUTÓNOMO EN VIVO ===\n")

for paso in range(1, 6):
    # Generamos lecturas aleatorias simuladas de los sensores entre 0.0 y 2.0 metros
    lectura_aleatoria = np.random.uniform(0.0, 2.0, size=(1, 3)).round(2)

    # La IA predice la acción buscando el patrón más cercano en su base de conocimientos
    prediccion = knn.predict(lectura_aleatoria)
    decision_texto = acciones[prediccion[0]]

    # Muestro los resultados del ciclo de control en tiempo real
    print(f"Pasos dados: {paso}")
    print(f"  📡 Sensores [Izq, Centro, Der]: {lectura_aleatoria[0]}")
    print(f"  🤖 Decisión de la IA:          {decision_texto}\n")
```
